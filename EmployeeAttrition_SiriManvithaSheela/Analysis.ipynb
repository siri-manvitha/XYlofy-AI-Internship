TASK-4:Model Building & Comparison


```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

import pandas as pd
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set:", X_train.shape)
print("Testing Set :", X_test.shape)
```

```python
# Scale only the training data
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])

X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])
```

```python
models = {
    "Logistic Regression":
        LogisticRegression(
            class_weight='balanced',
            random_state=42,
            max_iter=1000
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=100,
            class_weight='balanced',
            random_state=42
        ),

    "Gradient Boosting":
        # GradientBoostingClassifier(
        #     random_state=42
        # )
    GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)
}
```

```python
results = []

for name, model in models.items():

    # Train model
    model.fit(X_train, y_train)

    # Predictions
    y_pred = model.predict(X_test)
    
    y_prob = model.predict_proba(X_test)[:,1]

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    # Store results
    results.append({
    "Model": name,
    "Accuracy": accuracy_score(y_test,y_pred),
    "Precision": precision_score(y_test,y_pred),
    "Recall": recall_score(y_test,y_pred),
    "F1 Score": f1_score(y_test,y_pred),
    "ROC-AUC": roc_auc
})

    print("="*60)
    print(name)
    print("="*60)

    print(classification_report(y_test, y_pred))
```

```python
comparison = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ]
)

comparison = comparison.sort_values(
    by="ROC-AUC",
    ascending=False
)

comparison
```

Among the three evaluated models, Gradient Boosting demonstrated the best overall performance with the highest accuracy (85.03%) and ROC-AUC (0.8044), indicating superior discriminative capability. Logistic Regression achieved the highest recall (61.70%) and F1-score (44.27%), making it more suitable when maximizing the detection of positive cases is the priority. Although Random Forest achieved relatively high accuracy (83.67%), its very low recall (8.51%) suggests poor performance in identifying positive instances. Therefore, Gradient Boosting is recommended as the overall best-performing model, while Logistic Regression may be preferred in applications where minimizing false negatives is more important.

Task-7: HR Insights & Business Recommendations

Based on the feature importance analysis, the three factors that most strongly predict employee attrition are **OverTime**, **Monthly Income**, and **Years at Company** (or the top three features shown in your feature importance chart). Employees who work overtime frequently, earn lower salaries, or have fewer years of service are more likely to leave the company.

The **Sales department** should be the highest priority for retention efforts because it has the highest attrition rate. Within job roles, **Sales Representatives** and **Laboratory Technicians** experience the highest employee turnover and should receive focused attention from the HR team.

Salary alone does not explain employee attrition. While employees with lower monthly income are more likely to leave, factors such as overtime, job role, work-life balance, and years at the company also have a significant impact. This indicates that employee retention depends on multiple workplace factors rather than compensation alone.

Based on these findings, the company should implement two key HR strategies. First, identify employees who frequently work overtime and conduct regular retention discussions while promoting a healthier work-life balance. Second, provide career development opportunities, mentoring, and competitive compensation for employees in high-risk departments and job roles, particularly during their first few years with the company.

One limitation of this model is that it is trained on historical employee data and cannot capture personal reasons for leaving, such as family responsibilities, career goals, or external job opportunities. Therefore, HR should use the model as a decision-support tool rather than relying on it as the sole basis for employee retention decisions.
